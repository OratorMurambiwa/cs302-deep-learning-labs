**Part 1 – Import Libraries**

```python
import pandas as pd
from math import prod
```

**Part 2 – Upload and Load the Dataset**


```python
data = pd.read_csv("LabActivity2.csv")
data
```

**Part 3 – Understand the Dataset**

```python
print(data['Buys_Computer'].value_counts())
```

**Part 4 – Step 1: Compute Prior Probabilities**

```python
total=len(data)
print(total)

yes_count=(data['Buys_Computer']=='Yes').sum()
print(yes_count)

no_count=(data['Buys_Computer']=='No').sum()
print(no_count)

p_yes=yes_count/total
print(p_yes)

p_no=no_count/total
print(p_no)
```

**Part 5 – Step 2: Compute Conditional Probabilities**

```python
new_customer = {
    "Age": ">40",
    "Income": "Medium",
    "Student": "No",
    "Credit_Rating": "Excellent"
}
new_customer
```

```python
def conditional_probability(data, attribute, value, class_value):
    class_rows = data[data["Buys_Computer"] == class_value]
    matching_rows = class_rows[class_rows[attribute] == value]
    numerator = len(matching_rows)
    denominator = len(class_rows)
    probability = numerator / denominator
    return numerator, denominator, probability

yes_conditionals = {}
for attribute, value in new_customer.items():
    numerator, denominator, probability = conditional_probability(data, attribute, value, "Yes")
    yes_conditionals[attribute] = probability
    print(f"P({attribute}={value} | Yes) = {numerator}/{denominator} = {probability:.4f}")

no_conditionals = {}
for attribute, value in new_customer.items():
    numerator, denominator, probability = conditional_probability(data, attribute, value, "No")
    no_conditionals[attribute] = probability
    print(f"P({attribute}={value} | No) = {numerator}/{denominator} = {probability:.4f}")
```

**Part 6 – Step 3: Compute Likelihoods and Naïve Bayes Scores**

```python
likelihood_yes = prod(yes_conditionals.values())
likelihood_no = prod(no_conditionals.values())

print("Likelihood (Yes):", likelihood_yes)
print("Likelihood (No):", likelihood_no)

score_yes = p_yes * likelihood_yes
score_no = p_no * likelihood_no

print("P(X|Yes)=", round(likelihood_yes, 4))
print("P(X|No)=", round(likelihood_no, 4))

print("\n Naive Bayes Score for Class Label =yes", round(score_yes, 4))
print("\n Naive Bayes Score for Class Label =no", round(score_no, 4))
```

**Part 7 – Step 4: Compare Scores and Make the Prediction**

```python
if score_yes > score_no:
    prediction = "Yes"
elif score_no > score_yes:
    prediction = "No"
else:
    prediction = "Tie"

print("Predicted Class =", prediction)
```
